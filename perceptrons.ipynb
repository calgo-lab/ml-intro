## From Prototypes to Linear Classification

We classify $\mathbf{x}$ by comparing distances to class prototypes:

$\text{distance}(\mathbf{x}, \boldsymbol{\mu}_\Delta) > \text{distance}(\mathbf{x}, \boldsymbol{\mu}_o)$

$\|\mathbf{x} - \boldsymbol{\mu}_\Delta\| > \|\mathbf{x} - \boldsymbol{\mu}_o\|$

Squaring and expanding both sides:

$\mathbf{x}^\top\mathbf{x} - 2\boldsymbol{\mu}_\Delta^\top\mathbf{x} + \boldsymbol{\mu}_\Delta^\top\boldsymbol{\mu}_\Delta > \mathbf{x}^\top\mathbf{x} - 2\boldsymbol{\mu}_o^\top\mathbf{x} + \boldsymbol{\mu}_o^\top\boldsymbol{\mu}_o$

The $\mathbf{x}^\top\mathbf{x}$ terms cancel out, simplifying to:

$\boldsymbol{\mu}_\Delta^\top\mathbf{x} - \boldsymbol{\mu}_\Delta^2/2 < \boldsymbol{\mu}_o^\top\mathbf{x} - \boldsymbol{\mu}_o^2/2$

Rearranging:

$0 < (\boldsymbol{\mu}_o - \boldsymbol{\mu}_\Delta)^\top\mathbf{x} - \frac{1}{2}(\boldsymbol{\mu}_o^\top\boldsymbol{\mu}_o - \boldsymbol{\mu}_\Delta^\top\boldsymbol{\mu}_\Delta)$

We classify $\mathbf{x}$ by comparing distances to class prototypes:

$\text{distance}(\mathbf{x}, \boldsymbol{\mu}_\Delta) > \text{distance}(\mathbf{x}, \boldsymbol{\mu}_o)$

Rearranging: 
$$0 < (\boldsymbol{\mu}_o - \boldsymbol{\mu}_\Delta)^\top\mathbf{x} - \frac{1}{2}(\boldsymbol{\mu}_o^\top\boldsymbol{\mu}_o - \boldsymbol{\mu}_\Delta^\top\boldsymbol{\mu}_\Delta)$$


Setting $\mathbf{w} = \boldsymbol{\mu}_o - \boldsymbol{\mu}_\Delta$ and $\beta = \frac{1}{2}(\boldsymbol{\mu}_o^\top\boldsymbol{\mu}_o - \boldsymbol{\mu}_\Delta^\top\boldsymbol{\mu}_\Delta)$:

### Linear Classification

$\mathbf{w}^\top\mathbf{x} - \beta = \begin{cases} >0 & \text{if } \mathbf{x} \text{ belongs to class } \circ \\ <0 & \text{if } \mathbf{x} \text{ belongs to class } \Delta \end{cases}$

## Linear Classification: Finding Good Weights

**Question:** What is a good $\mathbf{w}$?

**Answer:** We need an **error function** that tells us how good $\mathbf{w}$ is.

## Two Classical Error Functions

Given data $\mathbf{x} \in \mathbb{R}^D$ and labels $y \in \{-1, +1\}$, two classical error functions to find optimal $\mathbf{w} \in \mathbb{R}^D$:

| Error Function | Used in |
|----------------|----------|
| $\frac{1}{2}(y - \mathbf{w}^\top\mathbf{x})^2$ | Adaline |
| $\max(0, -y\mathbf{w}^\top\mathbf{x})$ | Perceptron |

## The Perceptron Learning Algorithm: Goal

**Goal:** Binary classification of multivariate data $\mathbf{x} \in \mathbb{R}^D$

**Input:** 
- Learning rate $\eta$ 
- $N$ tuples $(\mathbf{x}_n, y_n)$ where:
  - $\mathbf{x}_n \in \mathbb{R}^D$ is the $D$-dimensional data
  - $y_n \in \{-1, +1\}$ is the corresponding label:

$y_n = \begin{cases} +1 & \text{if } \mathbf{x}_n \text{ is a correct example} \\ -1 & \text{if } \mathbf{x}_n \text{ is NOT a correct example} \end{cases}$

**Output:** Weight vector $\mathbf{w} \in \mathbb{R}^D$ such that:

$\mathbf{w}^\top\mathbf{x}_n = \begin{cases} \geq 0 & \text{if } y_n = +1 \\ < 0 & \text{if } y_n = -1 \end{cases}$

# Error Functions

## The Perceptron Error Function

The error function sums over all misclassified points:

$\mathcal{E}_P(\mathbf{w}) = -\sum_{m \in \mathcal{M}} \mathbf{w}^\top\mathbf{x}_m y_m$

where $\mathcal{M}$ is the set of indices of all **misclassified** data points $\mathbf{x}_m$.

## Classification Error as Function of Weights

<img src="figures/errorfunction_perceptron.pdf" width="400">

Given data $\mathbf{x} \in \mathbb{R}^D$ and labels $y \in \{-1, +1\}$, the classification error $\mathcal{E}$ is a function of the weights $\mathbf{w}$:

$\mathcal{E}(\mathbf{w}, \mathbf{x}_m, y_m) = -\sum_{m \in \mathcal{M}} \mathbf{w}^\top\mathbf{x}_m y_m$

where $\mathcal{M}$ denotes the index set of all **misclassified** data points.

# Gradient Descent

## Gradient Descent

<img src="figures/errorlandscape.pdf" width="450">

**Question:** How to minimize the error function?

$\mathcal{E}(\mathbf{w}, \mathbf{x}_m, y_m) = -\sum_{m \in \mathcal{M}} \mathbf{w}^\top\mathbf{x}_m y_m$

**Answer:** **Gradient Descent**

## Gradient Descent Algorithm

<img src="figures/errorlandscape.pdf" width="400">

We minimize $\mathcal{E}(\mathbf{w}, \mathbf{x}_m, y_m)$ by walking in the **opposite direction** of the gradient:

$\mathbf{w}^{\text{new}} \leftarrow \mathbf{w}^{\text{old}} - \eta \frac{1}{|\mathcal{X}|} \sum_{i=1}^{|\mathcal{X}|} \nabla\mathcal{E}(\mathbf{w}, \mathbf{x}_i, y_i)$

where:
- $\mathcal{X}$ is the set of data points
- $\eta$ is the **learning rate**

## Stochastic Gradient Descent

<img src="figures/errorlandscape.pdf" width="400">

A **noisy estimate** of the full gradient is obtained by using only one random data point:

$\mathbf{w}^{\text{new}} \leftarrow \mathbf{w}^{\text{old}} - \eta \nabla\mathcal{E}(\mathbf{w}, \mathbf{x}_i, y_i)$

**Advantages:**
- Only $\mathbf{w}$ needs to be stored
- Only one data point $\mathbf{x}_i$ and label $y_i$ are considered at a time
- **Scales to large datasets**

# Perceptron Learning

## Perceptron Training

Training a Perceptron means finding weights $\mathbf{w}$ that minimize $\mathcal{E}_P$:

$\underset{\mathbf{w}}{\arg\min} \left( -\sum_{m \in \mathcal{M}} \mathbf{w}^\top\mathbf{x}_m y_m \right)$

where $\mathcal{M}$ denotes the index set of all **misclassified** data $\mathbf{x}_m$.

## The Perceptron Learning Algorithm

The perceptron error can be minimized **iteratively** using **stochastic gradient descent**:

1. Initialize $\mathbf{w}^{\text{old}}$ (randomly, $1/D$, etc.)
2. While there are misclassified data points $\mathbf{x}_m$:
   - Pick a random misclassified data point $\mathbf{x}_m$
   - Update: $\mathbf{w}^{\text{new}} \leftarrow \mathbf{w}^{\text{old}} - \eta \nabla\mathcal{E}_P(\mathbf{w})$
   - Which simplifies to: $\mathbf{w}^{\text{new}} \leftarrow \mathbf{w}^{\text{old}} + \eta \mathbf{x}_m y_m$

## Problems with Perceptrons

- Each update might lead to new misclassifications
- Many solutions might exist; which one is correct?
- Convergence might be slow
- **Only solves linearly separable problems**
- If there is no solution, the algorithm will not converge

**Note:** Some solutions have been proposed (e.g., by Freund & Schapire, 1999)

## Multiclass Classification with Perceptrons

- Handwritten digit recognition is a **multiclass problem**
- But the Perceptron is a **binary classifier**
- **Solution:** One-vs-All strategy
  - Train a perceptron for each digit against all others
  - Concatenate all weight vectors: $\mathbf{W} = [\mathbf{w}_0, \mathbf{w}_1, ..., \mathbf{w}_9]$
  - For new data point $\mathbf{x}$, predict: $\underset{i}{\arg\max}(\mathbf{W}^\top\mathbf{x})$
