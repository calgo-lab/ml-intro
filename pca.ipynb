# Principal Component Analysis and Extensions


## Supervised vs Unsupervised Algorithms

- Often there is no label information available
  - Ongoing neural activity
  - Mixtures of different speakers in an audio recording
  - Complex artefacts in experimental recordings

- **Unsupervised** algorithms
  - Find structure in data sets
  - Allow partitioning of data in *meaningful* parts
  - Allow to remove unwanted aspects (e.g. noise)

## Principal Component Analysis

- Principal Component Analysis (PCA):

- Popular dimensionality reduction technique

- Easy to implement

## Maximizing Variance in a Data Set

We obtained some data $\mathbf{X}=[\mathbf{x}_1,\mathbf{x}_2,\dots,\mathbf{x}_N]\in\mathbb{R}^{D\times N}$

PCA finds a direction $\mathbf{w}^*\in\mathbb{R}^D$ such that

$$\mathbf{w}^* = \underset{\mathbf{w}}{\operatorname{argmax}} \; \mathbf{w}^{\top} \mathbf{X} \mathbf{X}^{\top} \mathbf{w}$$

When optimizing we have to constrain $\mathbf{w}$:

$$\|\mathbf{w}\|^2 = \mathbf{w}^{\top}\mathbf{w} = 1$$

yielding the Lagrangian

$$\mathcal{L} = \mathbf{w}^{\top}\mathbf{X}\mathbf{X}^{\top}\mathbf{w} + \lambda (1-\mathbf{w}^{\top}\mathbf{w})$$

## Short Excursion: Lagrangians and Optimization

> **Optimizing a function subject to some constraint**
>
> maximize $f(x,y)$
>
> subject to the constraint $g(x,y)=c$
>
> Lagrangian: $\mathcal{L}(x,y,\lambda) = f(x,y) + \lambda (g(x,y)-c)$
>
> where $\lambda$ is called a *Lagrangian Multiplier*

## Maximizing Variance in a Data Set

$$\mathcal{L} = \mathbf{w}^{\top}\mathbf{X}\mathbf{X}^{\top}\mathbf{w} + \lambda (1-\mathbf{w}^{\top}\mathbf{w})$$

Setting the derivative w.r.t. $\mathbf{w}$ to zero yields

$$\frac{\partial \mathcal{L}}{\partial \mathbf{w}} = 2 \mathbf{X}\mathbf{X}^{\top}\mathbf{w} - 2 \lambda \mathbf{w} = 0$$

$$\Rightarrow \mathbf{X}\mathbf{X}^{\top}\mathbf{w} = \lambda \mathbf{w}$$

This is a **standard eigenvalue problem**.

$\mathbf{w}$ is the eigenvector of $\mathbf{X}\mathbf{X}^{\top}$ corresponding to the **largest eigenvalue**

## Finding $k$ Principal Components

1. We found the strongest variance direction
2. Now we want to find the second strongest variance direction
3. The second direction should **not** be correlated with the first
4. We *project out* the variance explained by the first direction
5. And again find the strongest variance direction

How can we *project out* the variance along the first principal direction $\mathbf{w}_1$?

## Finding $k$ Principal Components: Deflation

Project data on $\mathbf{w}_1$ and back through $\mathbf{w}_1$:

$$\mathbf{X}_{\mathbf{w}_1} = \mathbf{w}_1 \underbrace{\mathbf{w}_1^\top \mathbf{X}}_{\text{First Principal Component}}$$

Now we can subtract $\mathbf{X}_{\mathbf{w}_1}$ from the original data $\mathbf{X}$:

$$\mathbf{X} - \mathbf{X}_{\mathbf{w}_1} = \mathbf{X} - \mathbf{w}_1 \mathbf{w}_1^\top \mathbf{X} = (\mathbf{I} - \mathbf{w}_1 \mathbf{w}_1^\top)\mathbf{X}$$

## Finding $k$ Principal Components: Projection

$$\mathbf{X}_{\mathbf{w}_1} = \mathbf{w}_1 \mathbf{w}_1^\top \mathbf{X}$$

## Finding $k$ Principal Components: Power Method

Finding the largest eigenvector using gradient descent, projecting it out and finding the next eigenvector is called the **Power Method** for solving eigenvalue equations

## Power Method for Eigendecompositions

**Algorithm: Power Method**

**Require:** Square matrix $\mathbf{A}$, number of eigenvector/eigenvalue pairs $k$

```
for k_i = 1 to k:
    # Initialize i-th eigenvector b randomly
    while not converged:
        b_i  <--  (A @ b_i) / || A @ b_i ||_2
    # Compute i-th eigenvalue
    lambda_i = A @ b_i
    # Deflate A
    A  <--  A - lambda_i * b_i @ b_i.T

return [b_1, ..., b_k]
```

## Principal Directions are Eigenvectors of Covariance Matrix

The $k$ first PCA basis vectors are the eigenvectors corresponding to the largest $k$ eigenvalues:

$$\mathbf{X}\mathbf{X}^{\top}\mathbf{W} = \mathbf{W} \Lambda$$

where $\mathbf{W}=[\mathbf{w}_1,\mathbf{w}_2,\dots,\mathbf{w}_k]$ contains the eigenvectors sorted according to their eigenvalues and $\Lambda$ is a diagonal matrix containing all eigenvalues.

A useful property of the new PCA basis — eigenvectors $\mathbf{w}_i,~i\in \{1,2,\dots,k\}$ are **orthogonal** to each other:

$$\mathbf{w}_i^{\top}\mathbf{w}_j = 0, \quad \forall i\neq j$$

## PCA Algorithm

**Algorithm: Principal Component Analysis**

**Require:** data $\mathbf{x}_1, \ldots, \mathbf{x}_N \in \mathbb{R}^d$, number of principal components $k$  
**Ensure:** $\mathbf{W}$

```
# Center Data
X = X - (1/N) * sum_i(x_i)

# Compute Covariance Matrix
C = (1/N) * X @ X.T

# Compute largest k eigenvectors
W = eig(C)
```

## Dimensionality Reduction by PCA

We can reduce the dimensionality of $\mathbf{X}$ from $d$ to $k$:

$$\mathbf{X}_{PCA} = \mathbf{W}^{\top}\mathbf{X}$$

If we want only a set $\mathcal{I}=\{i_1,i_2,\dots,i_n\}$ of principal components:

$$\mathbf{X}_{\mathcal{I}} = \sum_{i\in\mathcal{I}} \mathbf{w}_i \mathbf{w}_i^{\top} \mathbf{X}$$

Note that $\mathcal{I}$ does **not** need to contain the *strongest* components

## Denoising by PCA

Assuming that noise has high (or low) variance we can remove those components.

If we want to project out a set $\mathcal{I}=\{i_1,i_2,\dots,i_n\}$ of principal components but keep the data in the input space:

$$\mathbf{X}_{PCA} = \mathbf{X} - \mathbf{W}_{\mathcal{I}}\mathbf{W}_\mathcal{I}^{\top}\mathbf{X} = (\mathbf{I} - \mathbf{W}_{\mathcal{I}}\mathbf{W}_{\mathcal{I}}^{\top})\mathbf{X}$$

## PCA For High-Dimensional Data

We get a data set $\mathbf{X}=[\mathbf{x}_1,\mathbf{x}_2,\dots,\mathbf{x}_N]\in\mathbb{R}^{D\times N}$ where $N\ll D$

- $\rightarrow$ Covariance matrix $\mathbf{X}\mathbf{X}^{\top}$ will be very large ($D$-by-$D$)
- $\rightarrow$ Too few samples for a robust covariance matrix estimate

We know that $\mathbf{w}$ must lie in the span of the data:

$$\mathbf{w}=\mathbf{X} \mathbf{a}$$

where $\mathbf{a}$ is a weighting of each data point

## PCA For High-Dimensional Data: Kernel Trick

We can plug $\mathbf{w}=\mathbf{X} \mathbf{a}$ in the PCA objective and obtain

$$\mathbf{X}\underbrace{\mathbf{X}^{\top}\mathbf{X}}_{\text{Kernel } \mathbf{K}_X}\mathbf{a} = \lambda \mathbf{X} \mathbf{a}$$

which is equivalent to (Scholkopf et al., 1998):

$$\mathbf{K}_X \mathbf{a} = \lambda \mathbf{a}$$

Solving PCA via $\mathbf{X}^{\top}\mathbf{X}$ instead of $\mathbf{X}\mathbf{X}^{\top}$ is called **linear kernel PCA**

## Eigenvectors of $\mathbf{X}\mathbf{X}^{\top}$ and $\mathbf{X}^{\top}\mathbf{X}$

By Singular Value Decomposition we can decompose $\mathbf{X}$ into:

$$\mathbf{X} = \mathbf{E}\mathbf{S}\mathbf{F}^\top$$

Now we see that:

$$\text{Covariance: } \mathbf{X}\mathbf{X}^{\top} = \mathbf{E}\mathbf{S}^2\mathbf{E}^{\top}$$

$$\text{Kernel: } \mathbf{X}^{\top}\mathbf{X} = \mathbf{F}\mathbf{S}^2\mathbf{F}^{\top}$$

- $\rightarrow$ $\mathbf{E}$ are the eigenvectors of $\mathbf{X}\mathbf{X}^{\top}$
- $\rightarrow$ $\mathbf{F}$ are the eigenvectors of $\mathbf{X}^{\top}\mathbf{X}$
- $\rightarrow$ $\mathbf{S}$ are the (square root of the) eigenvalues of both
- $\rightarrow$ Relation linear kernel PCA and classical PCA: $\mathbf{E}\mathbf{S}=\mathbf{X}\mathbf{F}^{\top}$

## When to Use Which Form

- If there are more dimensions than samples ($N \ll D$)
  - $\rightarrow$ Compute PCA on linear kernel matrix $\mathbf{X}\mathbf{X}^{\top}\in\mathbb{R}^{N\times N}$

- If there are more samples than dimensions ($D \ll N$)
  - $\rightarrow$ Compute PCA on covariance matrix $\mathbf{X}^{\top}\mathbf{X}\in\mathbb{R}^{D\times D}$

## Summary

- **Unsupervised Data Analysis**
  - Finds structure in data in an explorative fashion
  - Can be used for dimensionality reduction, visualization, denoising

- **Principal Component Analysis (PCA)**
  - Finds directions of maximal variance
  - Is solved by eigendecomposition of Covariance/Kernel Matrix

- **Linear PCA**
  - Finds *linear* subspaces
  - If there are more dimensions than data points
    - $\rightarrow$ Do eigendecomposition on kernel matrix
